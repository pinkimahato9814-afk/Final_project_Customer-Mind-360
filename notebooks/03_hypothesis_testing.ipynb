## Hypothesis Testing
#### TWO_Sample test for numerical features


####  Question: Is the average Customer_Service_Calls different between churned and non-churned customers?

```python
alpha = 0.05
feature = 'Customer_Service_Calls'
```

```python
churned = df[df['Churned'] == 1][feature].dropna()
not_churned = df[df['Churned'] == 0][feature].dropna()
```

```python
t_statistics, p_value = stats.ttest_ind(
    churned,
    not_churned,
    equal_var=False
)
```

```python

h0= 'The average Customer_Service_Calls is the same for churned and non-churned customers.'
h1 = 'The average Customer_Service_Calls is different for churned and non-churned customers.'
print(f"H0: {h0}")
print(f"H1: {h1}")

print("Feature:", feature)
print("T-statistics:", t_statistics)
print("P-value:", p_value)
```

```python
if p_value > alpha:
    print(f'Decision: Failed to Reject null hypothesis.\nConclusion: Needs to accept Null Hypothesis.\n{h0}')
else:
    print(f'Decision: Failed to Accept null hypothesis.\nConclusion: Needs to accept Alternate Hypothesis.\n{h1}')
```

## chi-square Test

#### question:
Is customer churn associated with Signup_Quarter?

h0: Signup_Quarter and customer churn are independent.

h1: Signup_Quarter and customer churn are associated.

```python
alpha = 0.05
feature = 'Signup_Quarter'
```

```python
contingency_table = pd.crosstab(df[feature], df['Churned'])
```

```python
chi2_statistic, p_value, dof, expected = chi2_contingency(contingency_table)
```

```python
print("Feature:", feature)
print("Chi-square Statistic:", chi2_statistic)
print("P-Value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
if p_value > alpha:
    print(f'Decision: Failed to Reject null hypothesis.\nConclusion: Needs to accept Null Hypothesis.\n{h0}')
else:
    print(f'Decision: Failed to Accept null hypothesis.\nConclusion: Needs to accept Alternate Hypothesis.\n{h1}')
```
