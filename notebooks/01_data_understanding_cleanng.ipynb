## KPI calculation 

Key Performance Indicators (KPIs)

•	Total Customers = Count of all customers

•	Churn Rate = Churned Customers ÷ Total Customers × 100

•	Retention Rate = Non-Churned Customers ÷ Total Customers × 100

•	Average Lifetime Value = Average of Lifetime_Value

•	Average Order Value = Average of Average_Order_Value

•	Average Cart Abandonment Rate = Average of Cart_Abandonment_Rate

•	Average Return Rate = Average of Returns_Rate

•	Average Email Open Rate = Average of Email_Open_Rate

•	Average Login Frequency = Average of Login_Frequency

•	High-Risk Customers = Customers with high churn probability


```python
#Total Customers = Count of all customers

total_customers = df.shape[0]
churned_customers = df["Churned"].sum()
non_churned_customers = total_customers - churned_customers
churned_customers
```

```python
churn_rate = (churned_customers / total_customers) * 100
churn_rate
```

```python
retention_rate = (non_churned_customers / total_customers) * 100
retention_rate
```

```python
avg_lifetime_value = df["Lifetime_Value"].mean()
avg_lifetime_value
```

```python
avg_order_value = df["Average_Order_Value"].mean()
avg_order_value
```

```python
avg_return_rate = df["Returns_Rate"].mean()
avg_return_rate
```

```python

avg_email_open_rate = df["Email_Open_Rate"].mean()
avg_email_open_rate
```

```python

avg_login_frequency = df["Login_Frequency"].mean()
avg_login_frequency
```

### Churn by Age

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 35, 60, 100],
    # labels=["Child", "Teen", "Young Adult", "Adult", "Senior"]
     labels=["18-25", "26-35", "36-45", "46-55", "56+"]
)
```

```python
import warnings
warnings.filterwarnings("ignore")
# Calculate churn rate by age group
churn_by_age = (df.groupby("AgeGroup")["Churned"].mean() * 100)
churn_by_age
```

```python
df["Churned"] = df["Churned"].replace({
    "Yes": 1,
    "No": 0,
    "yes": 1,
    "no": 0
})
```

```python
# Plot bar chart
plt.figure(figsize=(8, 5))

plt.bar(
    churn_by_age.index.astype(str),
    churn_by_age.values,
    color="purple"
)

plt.title("Churn Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Churn Rate (%)")

plt.ylim(0, 100)

for i, value in enumerate(churn_by_age.values):
    plt.text(i, value + 1, f"{value:.1f}%", ha="center")

plt.show()
```

## 3. Churn by Country or City
- A bar chart showing churn count or churn rate by country and city.


```python

country_city_churn = df.groupby(["Country", "City"]).agg(
     total_customers=("Churned", "count"),
    churned_customers=("Churned", "sum"),
    churn_rate=("Churned", "mean")
).reset_index()
# Convert churn rate into percentage
country_city_churn["churn_rate"] = country_city_churn["churn_rate"] * 100
# Sort by churned customers
country_city_churn = country_city_churn.sort_values(
    by="churned_customers",
    ascending=False
)

country_city_churn.head(10)
```

```python
import matplotlib.pyplot as plt

top_country_city = country_city_churn.head(10).copy()

top_country_city["Country_City"] = (
    top_country_city["Country"].astype(str) + " - " + top_country_city["City"].astype(str)
)

plt.figure(figsize=(12, 5))

plt.bar(
    top_country_city["Country_City"],
    top_country_city["churned_customers"],
    color = 'purple'
)

plt.title("Top 10 Country-City by Churn Count")
plt.xlabel("Country - City")
plt.ylabel("Number of Churned Customers")
plt.xticks(rotation=45, ha="right")

# Add values on top of bars
for i, value in enumerate(top_country_city["churned_customers"]):
    plt.text(i, value + 1, f"{value:.0f}", ha="center")

plt.show()
```

## 4. Membership Years vs Churn

```python

plt.figure(figsize=(8, 5))

plt.bar(
    membership_churn.index.astype(str),
    membership_churn.values,
    color = ['green','red'],
    width =0.4
)

plt.title("Average Membership Years by Churn Status")
plt.xlabel("Churn Status")
plt.ylabel("Average Membership Years")

plt.xticks([0, 1], ["Not Churned", "Churned"])

# # Zoom y-axis around the actual values
plt.ylim(2.9, 3.05)

for i, value in enumerate(membership_churn.values):
    plt.text(i, value + 0.005, f"{value:.2f}", ha="center")

plt.show()
```

The average membership years of churned and non-churned customers are almost the same. This means Membership_Years alone may not strongly explain customer churn in this dataset.

## 5. Login Frequency vs Churn
- A box plot or histogram comparing login frequency for churned and non-churned customers.


```python
LG_frequency_churn = df.groupby('Churned')['Login_Frequency'].mean()

plt.figure(figsize=(8, 5))

plt.bar(
    LG_frequency_churn.index,
    LG_frequency_churn.values,
    color=['green', 'red'],  width=0.6
)

plt.title("Average Login Frequency by Churn Status",fontsize=14, pad=15)
plt.xlabel("Customer Status",fontsize=12)
plt.ylabel("Average Login Frequency",fontsize=12)

plt.xticks([0, 1], ["Not Churned", "Churned"])


# Add extra space above the highest bar
plt.ylim(0, LG_frequency_churn.max() + 5)

# Add value labels on top of bars
for i, value in enumerate(LG_frequency_churn.values):
    plt.text(
        i,
        value + 0.3,
        f"{value:.2f}",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

plt.tight_layout()
plt.show()
```

The chart shows that non-churned customers have a higher average login frequency than churned customers.
This indicates that customers who log in more frequently are more engaged and less likely to churn.
