## KPI calculation 

Key Performance Indicators (KPIs)

•	Total Customers = Count of all customers

•	Churn Rate = Churned Customers ÷ Total Customers × 100

•	Retention Rate = Non-Churned Customers ÷ Total Customers × 100

•	Average Lifetime Value = Average of Lifetime_Value

•	Average Order Value = Average of Average_Order_Value

•	Average Cart Abandonment Rate = Average of Cart_Abandonment_Rate

•	Average Return Rate = Average of Returns_Rate

•	Average Email Open Rate = Average of Email_Open_Rate

•	Average Login Frequency = Average of Login_Frequency

•	High-Risk Customers = Customers with high churn probability


```python
#Total Customers = Count of all customers

total_customers = df.shape[0]
churned_customers = df["Churned"].sum()
non_churned_customers = total_customers - churned_customers
churned_customers
```

```python
churn_rate = (churned_customers / total_customers) * 100
churn_rate
```

```python
retention_rate = (non_churned_customers / total_customers) * 100
retention_rate
```

```python
avg_order_value = df["Average_Order_Value"].mean()
avg_order_value
```

## 3. Churn by Country or City
- A bar chart showing churn count or churn rate by country and city.


```python

country_city_churn = df.groupby(["Country", "City"]).agg(
     total_customers=("Churned", "count"),
    churned_customers=("Churned", "sum"),
    churn_rate=("Churned", "mean")
).reset_index()
# Convert churn rate into percentage
country_city_churn["churn_rate"] = country_city_churn["churn_rate"] * 100
# Sort by churned customers
country_city_churn = country_city_churn.sort_values(
    by="churned_customers",
    ascending=False
)

country_city_churn.head(10)
```

```python
import matplotlib.pyplot as plt

top_country_city = country_city_churn.head(10).copy()

top_country_city["Country_City"] = (
    top_country_city["Country"].astype(str) + " - " + top_country_city["City"].astype(str)
)

plt.figure(figsize=(12, 5))

plt.bar(
    top_country_city["Country_City"],
    top_country_city["churned_customers"],
    color = 'purple'
)

plt.title("Top 10 Country-City by Churn Count")
plt.xlabel("Country - City")
plt.ylabel("Number of Churned Customers")
plt.xticks(rotation=45, ha="right")

# Add values on top of bars
for i, value in enumerate(top_country_city["churned_customers"]):
    plt.text(i, value + 1, f"{value:.0f}", ha="center")

plt.show()
```

## 10. Discount Usage Rate Analysis
- A bar chart or box plot comparing discount usage with churn.

```python
df['Churned'].unique()
```

```python
# Convert Churned column into numeric if needed
df["Churned"] = df["Churned"].replace({
    "Yes": 1,
    "No": 0,
})
```

```python
df["Churned"] = pd.to_numeric(df["Churned"], errors="coerce")# put numeric values in churned column for further analysis    
```

```python
#Clean Discount_Usage_Rate
df["Discount_Usage_Rate"] = pd.to_numeric(df["Discount_Usage_Rate"], errors="coerce")
```

```python
df["Discount_Usage_Rate"].max()
```

```python
# Create discount usage groups
df["Discount_Usage_Group"] = pd.cut(
    df["Discount_Usage_Rate"],
    bins=[0, 33, 66, 100],
    labels=["Low Discount Usage", "Medium Discount Usage", "High Discount Usage"],
    include_lowest=True
)
```

```python

# Group by Discount Usage Group and Churned
discount_churn = df.groupby(["Discount_Usage_Group",'Churned']).size().unstack()
discount_churn
```

```python
# Rename columns
discount_churn.columns = ["Not Churned", "Churned"]
```

```python

# Plot column bar chart
discount_churn.plot(
    kind="bar",
    figsize=(10, 6),
    color = ['lightgreen', 'hotpink'],
)
plt.title("Discount Usage Rate Analysis by Churn Status")
plt.xlabel("Discount Usage Group")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.legend(title="Customer Status")
# Add value labels
for container in plt.gca().containers:
    plt.bar_label(container)

plt.show()

```

```python

# Calculate churn rate by discount usage group
discount_churn_rate = df.groupby("Discount_Usage_Group")["Churned"].mean() * 100
```

```python

plt.figure(figsize=(8, 5))
plt.bar(
    discount_churn_rate.index.astype(str),
    discount_churn_rate.values,
    width=0.5,
    color=['red', 'orange', 'lightblue']
)

plt.title("Churn Rate by Discount Usage Group")
plt.xlabel("Discount Usage Group")
plt.ylabel("Churn Rate (%)")
plt.ylim(0, 100)

for i, value in enumerate(discount_churn_rate.values):
    plt.text(i, value + 1, f"{value:.2f}%", ha="center")

plt.show()
```

<mark>Report:<mark>

- If Low Discount Usage then 34.84% Churned rate .
- If Medium Discount Usage then 25.89% Churned rate.
- If High Discount Usage then 25.96% Churned rate .
