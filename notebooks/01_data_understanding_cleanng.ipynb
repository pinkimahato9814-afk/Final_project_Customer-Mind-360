```python
plt.hist(df['Session_Duration_Avg'],bins =20,edgecolor = 'black')
```

```python
mean_SAD= df['Session_Duration_Avg'].median()
mean_SAD_rounded = round(mean_SAD,0)
df['Session_Duration_Avg']  =df['Session_Duration_Avg'].fillna(mean_SAD_rounded)
```

```python
plt.hist(df['Session_Duration_Avg'],bins =20,edgecolor = 'black')
```

```python
plt.hist(df['Pages_Per_Session'],bins = 20,edgecolor = 'black')
```

```python
median_val = df['Pages_Per_Session'].median()
rnd_median_val = round(median_val,0)
df['Pages_Per_Session'] = df['Pages_Per_Session'].fillna(rnd_median_val)
```

```python
plt.hist(df['Wishlist_Items'],bins =20,edgecolor = 'black')
```

```python
median_val =df['Wishlist_Items'].median()
df['Wishlist_Items'] = df['Wishlist_Items'].fillna(median_val)
```
